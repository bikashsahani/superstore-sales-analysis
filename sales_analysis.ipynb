9. Are there outliers in Sales, Profit, or Quantity?

```python
# function to check and handle outliers
def handle_outliers(df, column_name):
    # 1. Calculate IQR statistics using clean, local variable names
    Q1 = df[column_name].quantile(0.25)
    Q3 = df[column_name].quantile(0.75)
    IQR = Q3 - Q1
    
    # 2. Define bounds
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # 3. Find and count outliers
    outliers_mask = (df[column_name] < lower_bound) | (df[column_name] > upper_bound)
    outlier_count = outliers_mask.sum()
    print(f"Found {outlier_count} outliers in '{column_name}'")
    
    # # 4. Replace outliers with the column median
    # median_val = df[column_name].median()
    # df.loc[outliers_mask, column_name] = median_val
    
    return df
```

```python
data = handle_outliers(data, 'Sales')
data = handle_outliers(data, 'Profit')
data = handle_outliers(data, 'Quantity')
```

📅 7. Time-Based Analysis

27. What is the monthly sales trend over time?

```python
# Create a 'Year-Month' column for grouping (e.g., '2026-06')
data['Year-Month'] = data['Order Date'].dt.to_period('M')

# Group by the year-month period and sum the sales
monthly_sales = data.groupby('Year-Month')['Sales'].sum().reset_index()

# Convert 'Year-Month' back to timestamp format (makes it easier for plotting tools)
monthly_sales['Year-Month'] = monthly_sales['Year-Month'].dt.to_timestamp()

print(monthly_sales)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

sns.lineplot(data=monthly_sales, x='Year-Month', y='Sales', marker='o', color='b', linewidth=2)

plt.title('Monthly Sales Trend Over Time', fontsize=14, fontweight='bold')
plt.xlabel('Date (Monthly)', fontsize=12)
plt.ylabel('Total Sales ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)

# Display the plot
plt.show()
```

33. Do a small number of customers contribute most of the revenue (Pareto 80/20 rule)?

```python
# 1. Group by Customer and calculate total sales
customer_sales = data.groupby('Customer ID')['Sales'].sum().reset_index()

# 2. Sort customers from highest revenue to lowest
customer_sales = customer_sales.sort_values(by='Sales', ascending=False).reset_index(drop=True)

# Total sales across the entire dataset
total_revenue = customer_sales['Sales'].sum()
total_customers = len(customer_sales)

# 1. Cumulative sales percentage
customer_sales['Cumulative Sales'] = customer_sales['Sales'].cumsum()
customer_sales['Cumulative Sales %'] = (customer_sales['Cumulative Sales'] / total_revenue) * 100

# 2. Cumulative customer percentage
customer_sales['Customer Rank %'] = ((customer_sales.index + 1) / total_customers) * 100

print(customer_sales[['Customer ID', 'Sales', 'Customer Rank %', 'Cumulative Sales %']].head(10))
```

```python
# Filter for the top 20% of customers
top_20_percent_customers = customer_sales[customer_sales['Customer Rank %'] <= 20]

# Get the maximum cumulative sales percentage within this group
revenue_share = top_20_percent_customers['Cumulative Sales %'].max()

print(f"The top 20% of customers contribute {revenue_share:.2f}% of the total revenue.")
```
